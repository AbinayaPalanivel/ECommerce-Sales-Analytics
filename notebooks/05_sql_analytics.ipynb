# Phase 5: SQL Analytics

## Objective

SQL is used to analyze structured business data efficiently.

In this notebook, we will:

- Import the cleaned dataset into SQLite
- Execute SQL queries
- Analyze sales, profit, customers, and regions
- Compare SQL analysis with Python analysis

### Tools Used

- SQLite
- SQL
- Python
- Pandas

```python
import pandas as pd
import sqlite3

print("Libraries Loaded Successfully!")
```

## Load the Cleaned Dataset

Load the cleaned CSV file created during the Data Cleaning phase.

```python
df = pd.read_csv("../data/cleaned_superstore.csv")

df.head()
```

## Create SQLite Database

Create a local SQLite database and import the dataset into a table named `orders`.

```python
conn = sqlite3.connect("ecommerce.db")

df.to_sql(
    "orders",
    conn,
    if_exists="replace",
    index=False
)

print("Database Created Successfully!")
```

## Query 1 – Total Sales

```python
query = """
SELECT
ROUND(SUM(Sales),2) AS Total_Sales
FROM orders;
"""

pd.read_sql(query, conn)
```

## Query 2 – Total Profit

```python
query = """
SELECT
ROUND(SUM(Profit),2) AS Total_Profit
FROM orders;
"""

pd.read_sql(query, conn)
```

## Query 3 – Sales by Category

```python
query = """
SELECT
Category,
ROUND(SUM(Sales),2) AS Sales
FROM orders
GROUP BY Category
ORDER BY Sales DESC;
"""

pd.read_sql(query, conn)
```

## Query 4 – Sales by Region

```python
query = """
SELECT
Region,
ROUND(SUM(Sales),2) AS Sales
FROM orders
GROUP BY Region
ORDER BY Sales DESC;
"""

pd.read_sql(query, conn)
```

## Query 5 – Top 10 Customers

```python
query = """
SELECT
`Customer Name`,
ROUND(SUM(Sales),2) AS Total_Sales
FROM orders
GROUP BY `Customer Name`
ORDER BY Total_Sales DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

## Query 6 – Top 10 Products

```python
query = """
SELECT
`Product Name`,
ROUND(SUM(Sales),2) AS Sales
FROM orders
GROUP BY `Product Name`
ORDER BY Sales DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

## Close Database Connection

Always close the database connection after completing SQL operations.

```python
conn.close()

print("Database Connection Closed!")
```
