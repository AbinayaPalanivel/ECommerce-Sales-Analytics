# Phase 2: Data Cleaning
Objective
Before analyzing the dataset, we need to clean it to improve data quality.
Tasks
- Load the dataset
- Check for duplicate records
- Remove duplicates (if any)
- Check for missing values
- Verify data types
- Save the cleaned dataset

```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_excel("../data/sample_-_superstore.xls")

print("Dataset loaded successfully!")
```

## Check Dataset Shape

This helps us compare the dataset size before and after cleaning.

```python
print("Dataset Shape:", df.shape)
```

## Check for Duplicate Records

Duplicate rows can affect analysis. We count them before deciding whether to remove them.

```python
duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)
```

## Remove Duplicate Records

If duplicates exist, remove them and keep the first occurrence.

```python
df = df.drop_duplicates()

print("New Shape:", df.shape)
```

## Check Missing Values

Identify columns containing null values.

```python
df.isnull().sum()
```

## Verify Data Types

Check whether each column has the correct data type.

```python
df.dtypes
```

## Save the Cleaned Dataset

Save the cleaned dataset for later notebooks.

```python
df.to_csv("../data/cleaned_superstore.csv", index=False)

print("Cleaned dataset saved successfully!")
```
